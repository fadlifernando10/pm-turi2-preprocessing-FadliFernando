```python
import pandas as pd
import numpy as np

data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],
    'divisi'    : ['IT', 'HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT'],
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']
}
df = pd.DataFrame(data)
df
```

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].mean())
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})
X = pd.get_dummies(X, columns=['kota', 'divisi'], dtype=int)
y = y.map({'Tidak': 0, 'Ya': 1})
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
from sklearn.preprocessing import MinMaxScaler

num = ['usia', 'pendapatan', 'pendidikan']
sc = MinMaxScaler()

X_train[num] = sc.fit_transform(X_train[num])
display(X_train.describe())
```

```python
print("Jumlah missing value:", X_train.isnull().sum().sum())
display(X_train.head())
```

2. Kapan label encoding dan kapan one-hot encoding?

label encoding: digunakan jika data kategori memiliki tingkatan atau urutan yang jelas. contohnya: sma - s1 - s2

one-hot encodingdigunakan jika data kategori nominal yang tidak punya tingkatan. contoh: nama kota seperti bandung, jakarta, surabaya agar komputer tidak salah menganggap angka kategorinya memiliki urutan nilai

3. Apa perbedaan normalisasi dan standarisasi?

normalisasi seperti MinMaxScaler mengubah skala data ke rentang tertentu(0 sampai 1) sedangkan standarisasi seperti StandardScaler mengubah data agar memiliki rata rata (mean) sama dengan 0 dan standar deviasi sama dengan 1
